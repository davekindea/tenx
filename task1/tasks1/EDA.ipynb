```python

range_values = []
counts = []
skews = []
kurtosis_data1 = []

for col in data1.columns:
    if data1[col].dtype in ['int64', 'float64']:  
        col_range = data1[col].max() - data1[col].min()
        range_values.append(col_range)
        
        col_count = data1[col].count()
        counts.append(col_count)
        
        col_skew = data1[col].skew()
        skews.append(col_skew)
        

        col_kurtosis = data1[col].kurtosis()
        kurtosis_data1.append(col_kurtosis)
summary_df = pd.DataFrame({
    'Column': data1.select_dtypes(include=['int64', 'float64']).columns,
    'Range': range_values,
    'Count': counts,
    'Skewness': skews,
    'Kurtosis': kurtosis_data1
   
})


print(summary_df)
```

```python
from scipy import stats

for col in ['GHI', 'DNI', 'DHI', 'Tamb']:
    data1[f'{col}_zscore'] = stats.zscore(data1[col])
    anomalies = data1[data1[f'{col}_zscore'].abs() > 3]
    plt.figure(figsize=(15, 6))
    plt.plot(data1.index, data1[col], label='Normal')
    plt.scatter(anomalies.index, anomalies[col], color='red', label='Anomaly')
    plt.title(f'Anomaly Detection in {col}')
    plt.legend()
    plt.show()
```

```python
before_cleaning=data1[data1["Cleaning"]==0]
after_cleaning=data1[data1["Cleaning"]==1]

print(f" Before Cleaning {before_cleaning.shape[0]} rows and {before_cleaning.shape[1]} Columns")
print(f" After cLeaning {after_cleaning.shape[0]} rows and {before_cleaning.shape[1]} Columns")
```

```python
plt.figure(figsize=(15, 6))

# Plot ModA for before and after cleaning
plt.subplot(2, 1, 1)
plt.plot(before_cleaning.index, before_cleaning['ModA'], label='Before Cleaning', color='blue')
plt.plot(after_cleaning.index, after_cleaning['ModA'], label='After Cleaning', color='green')
plt.title('ModA Sensor Readings Before and After Cleaning')
plt.xlabel('Time')
plt.ylabel('ModA')
plt.legend()

# Plot ModB for before and after cleaning
plt.subplot(2, 1, 2)
plt.plot(before_cleaning.index, before_cleaning['ModB'], label='Before Cleaning', color='blue')
plt.plot(after_cleaning.index, after_cleaning['ModB'], label='After Cleaning', color='green')
plt.title('ModB Sensor Readings Before and After Cleaning')
plt.xlabel('Time')
plt.ylabel('ModB')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import ttest_ind

# Perform a t-test between ModA before and after cleaning
t_stat_modA, p_value_modA = ttest_ind(before_cleaning['ModA'], after_cleaning['ModA'])
print(f"T-test for ModA: t-statistic = {t_stat_modA}, p-value = {p_value_modA}")

# Perform a t-test between ModB before and after cleaning
t_stat_modB, p_value_modB = ttest_ind(before_cleaning['ModB'], after_cleaning['ModB'])
print(f"T-test for ModB: t-statistic = {t_stat_modB}, p-value = {p_value_modB}")
```

```python
# Compute moving average (e.g., 7-day window) for ModA and ModB
before_cleaning['ModA_MA'] = before_cleaning['ModA'].rolling(window=7).mean()
after_cleaning['ModA_MA'] = after_cleaning['ModA'].rolling(window=7).mean()

before_cleaning['ModB_MA'] =before_cleaning['ModB'].rolling(window=7).mean()
after_cleaning['ModB_MA'] = after_cleaning['ModB'].rolling(window=7).mean()

# Plot moving averages
plt.figure(figsize=(15, 6))

# ModA moving average
plt.subplot(2, 1, 1)
plt.plot(before_cleaning.index, before_cleaning['ModA_MA'], label='Before Cleaning', color='blue')
plt.plot(after_cleaning.index, after_cleaning['ModA_MA'], label='After Cleaning', color='green')
plt.title('ModA Sensor Readings with Moving Average Before and After Cleaning')
plt.xlabel('Time')
plt.ylabel('ModA (Moving Average)')
plt.legend()

# ModB moving average
plt.subplot(2, 1, 2)
plt.plot(before_cleaning.index, before_cleaning['ModB_MA'], label='Before Cleaning', color='blue')
plt.plot(after_cleaning.index, after_cleaning['ModB_MA'], label='After Cleaning', color='green')
plt.title('ModB Sensor Readings with Moving Average Before and After Cleaning')
plt.xlabel('Time')
plt.ylabel('ModB (Moving Average)')
plt.legend()

plt.tight_layout()
plt.show()
```
